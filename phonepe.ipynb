```python
import os
import json
import pandas as pd
```

```python
# Folder path for Transaction data
transaction_path = r"C:\Users\Varshini V\OneDrive\Documents\VS_code\pulse\data\aggregated\transaction\country\india\state/"

states = os.listdir(transaction_path) # extracts the states from the transaction path and stores in the states parameter aas list
transaction_list = []

for state in states:
    state_path = transaction_path + state + "/" # extractes the path including the state names
    years = os.listdir(state_path) # creates the list of years of each state
    
    for year in years:
        year_path = state_path + year + "/" #extractes the list of years from the years parameter and converts to path
        quarters = os.listdir(year_path) #creates the list of quarters for each year

        for quarter in quarters:
            file_path = year_path + quarter
            with open(file_path) as f:
                data = json.load(f)

             # Extract data safely
            if data["data"]["transactionData"] is not None:
                for item in data["data"]["transactionData"]:
                    for instrument in item["paymentInstruments"]:
                        transaction_list.append({
                            "State": state,
                            "Year": year,
                            "Quarter": quarter.strip(".json"),
                            "Transaction_Name": item["name"],
                            "Transaction_Count": instrument["count"],
                            "Transaction_Amount": instrument["amount"]
                        })

# Convert to DataFrame
df_transaction = pd.DataFrame(transaction_list)
df_transaction
```

```python
# Folder path for User data
user_path = r"C:\Users\Varshini V\OneDrive\Documents\VS_code\venv\program\pulse\data\aggregated\user\country\india\state/"

states = os.listdir(user_path)
user_list = []

for state in states:
    state_path = user_path + state + "/"
    years = os.listdir(state_path)

    for year in years:
        year_path = state_path + year + "/"
        quarters = os.listdir(year_path)

        for quarter in quarters:
            file_path = year_path + quarter
            with open(file_path) as f:
                data = json.load(f)

            if data["data"]["usersByDevice"] is not None:
                for item in data["data"]["usersByDevice"]:
                    user_list.append({
                        "State": state,
                        "Year": year,
                        "Quarter": quarter.strip(".json"),
                        "Device_Brand": item["brand"],
                        "Device_Count": item["count"],
                        "Device_Percentage": item["percentage"]
                    })

# Convert to DataFrame
df_user = pd.DataFrame(user_list)
df_user
```

```python
# Folder path for User data
user_map = r"C:\Users\Varshini V\OneDrive\Documents\VS_code\venv\program\pulse\data\map\user\hover\country\india\state/"

states = os.listdir(user_map)
user_list2 = []

for state in states:
    state_path = os.path.join(user_map, state)
    years = os.listdir(state_path)

    for year in years:
        year_path = os.path.join(state_path, year)
        quarters = os.listdir(year_path)

        for quarter in quarters:
            file_path = os.path.join(year_path, quarter)
            with open(file_path) as f:
                data = json.load(f)

            if data.get("data") and "hoverData" in data["data"]:
                for district, values in data["data"]["hoverData"].items():
                    user_list2.append({
                        "State": state,
                        "Year": year,
                        "Quarter": quarter.strip(".json"),
                        "District": district,
                        "Registered_users": values.get("registeredUsers"),
                        "App_opens": values.get("appOpens")
                    })

# Convert to DataFrame
df_user_map = pd.DataFrame(user_list2)
print(df_user_map)
```
